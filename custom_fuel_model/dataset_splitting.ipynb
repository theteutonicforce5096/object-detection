```python
import os
import shutil

from ultralytics.data.split import autosplit
```

```python
# Annotated dataset needs to be organized into two folders: images and labels
autosplit(
    path = "./data/annotated/images",
    weights = (0.8, 0.2, 0.0),
    annotated_only = True
)
```

```python
# Create the directories that will house the train and valid images and labels
os.makedirs("data/annotated/train/images", exist_ok=True)
os.makedirs("data/annotated/train/labels", exist_ok=True)
os.makedirs("data/annotated/valid/images", exist_ok=True)
os.makedirs("data/annotated/valid/labels", exist_ok=True)

# Move the corresponding images and labels to the train and valid folders
with open("./data/annotated/autosplit_train.txt", "r") as f:
    file_paths = f.read().splitlines()

    for file_path in file_paths:
        rel_image_file_path = "data/annotated/images" + file_path[8:]
        rel_label_file_path = "data/annotated/labels" + file_path[8:-3] + "txt"
        shutil.move(rel_image_file_path, "data/annotated/train/images")
        shutil.move(rel_label_file_path, "data/annotated/train/labels")

with open("./data/annotated/autosplit_val.txt", "r") as f:
    file_paths = f.read().splitlines()

    for file_path in file_paths:
        rel_image_file_path = "data/annotated/images" + file_path[8:]
        rel_label_file_path = "data/annotated/labels" + file_path[8:-3] + "txt"
        shutil.move(rel_image_file_path, "data/annotated/valid/images")
        shutil.move(rel_label_file_path, "data/annotated/valid/labels")
```

```python
# Clean up old directories and files
os.rmdir("data/annotated/images")
os.rmdir("data/annotated/labels")
os.remove("autosplit_train.txt")
os.remove("autosplit_val.txt")
```
