```python
from ultralytics import YOLO
```

```python
# Load the pre-trained YOLOv11s model
model = YOLO("yolo11s.pt")  # load a pretrained model (recommended for training)

# Train the model with MPS (Apple Silicon only)
results = model.train(
    data="data/annotated/data.yaml", 
    epochs=100, 
    patience = 0, 
    imgsz=640, 
    device="mps", 
    # quantize = "int8",
    # batch = 8,
    cache = "ram",
    # workers = 4
)
```
